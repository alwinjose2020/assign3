```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
df = pd.read_csv("customer_behavior.csv")
print(df.head())

print("Mean:", df["PurchaseAmount"].mean())
print("Median:", df["PurchaseAmount"].median())
print("Mode:", df["PurchaseAmount"].mode().tolist())
sns.boxplot(x=df["PurchaseAmount"])
plt.show()
```

```python
print("Skewness:", df["PurchaseAmount"].skew())
print("Kurtosis:", df["PurchaseAmount"].kurtosis())
```

```python
male = df[df["Gender"] == "Male"]["PurchaseAmount"]
female = df[df["Gender"] == "Female"]["PurchaseAmount"]
print("Male Average:", male.mean())
print("Female Average:", female.mean())
print(stats.ttest_ind(male, female, equal_var=False))
```
